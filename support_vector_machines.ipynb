## SVM Kernels Indepth and Practical Explanation

```python
x = np.linspace(-5.0, 5.0, 100)
y = np.sqrt(10**2 - x**2)
x = np.hstack([x, -x])
y = np.hstack([y, -y])
```

```python
x1 = np.linspace(-5.0, 5.0, 100)
y1 = np.sqrt(5**2 - x1**2)
x1 = np.hstack([x1, -x1])
y1 = np.hstack([y1, -y1])
```

```python
plt.scatter(y, x)
plt.scatter(y1, x1)
```

```python
np.vstack([y, x]).T
```

```python
df1 = pd.DataFrame(np.vstack([y, x]).T, columns=['X1', 'X2'])
df1['Y'] = 0
df2 = pd.DataFrame(np.vstack([y1, x1]).T, columns=['X1', 'X2'])
df2['Y'] = 1
df = pd.concat([df1, df2], ignore_index=True)
df.sample(5)
```

```python
# Independent and Dependent features
x = df.iloc[:, :2]
y = df.Y
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

```python
y_train
```

```python
# Polynomial Kernel:- K(x, y) = ((x^T)*y + c)^d
```

```python
# We need to find components for the Polynomial Kernel
# X1, X2, X1_Square, X2_Square, X1*X2

df['X1_Square'] = df['X1']**2
df['X2_Square'] = df['X2']**2
df['X1*X2'] = df['X1'] * df['X2']
df.sample(5)
```

```python
# Independent and Dependent features
x = df[['X1', 'X2', 'X1_Square', 'X2_Square', 'X1*X2']]
y = df['Y']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=0)
```

```python
# Independent and Dependent Feature
x=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y = df['total_bill']
```

```python
x
```

```python
y
```

```python
# train test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=10)
```

#### Feature Encoding (Label Encoding and Onehot Encoding)

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
```

```python
x_train['sex']=le1.fit_transform(x_train['sex'])
x_train['smoker']=le2.fit_transform(x_train['smoker'])
x_train['time']=le3.fit_transform(x_train['time'])
```

```python
x_train.head()
```

```python
x_test['sex']=le1.transform(x_test['sex'])
x_test['smoker']=le2.transform(x_test['smoker'])
x_test['time']=le3.transform(x_test['time'])
```

```python
# Onehot Encoding---- apply ColumnTransformer

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(transformers=[('onehot', OneHotEncoder(drop='first'), [3])], remainder='passthrough')
```

```python
import sys
np.set_printoptions(threshold=sys.maxsize)
x_train=ct.fit_transform(x_train)
```

```python
x_test = ct.transform(x_test)
```

```python
x_test
```

```python
# Apllying SVR---Support Vector Regression
from sklearn.svm import SVR
svr = SVR()
```

```python
svr.fit(x_train, y_train)
```

```python
y_pred = svr.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
print(r2_score(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
```

#### Hyperparameter Tunning using GridSearchCV

```python
# defining parameter range
param_grid = {'C' : [0.1, 1, 10, 100, 1000],
                'gamma' : [1, 0.1, 0.01, 0.001, 0.0001],
                'kernel' : ['rbf']}
```

```python
grid = GridSearchCV(SVR(), param_grid, refit=True, verbose=3)
# fitting the model for grid search
grid.fit(x_train, y_train)
```

```python
grid.best_params_
```

```python
grid_prediction = grid.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
print(r2_score(y_test, grid_prediction))
print(mean_absolute_error(y_test, grid_prediction))
```
